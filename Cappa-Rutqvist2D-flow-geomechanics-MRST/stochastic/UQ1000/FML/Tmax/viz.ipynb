```python
def log_min_max_scale(X,select_idx):
    logX = np.log(X)
    X_train = logX[:,:,select_idx]
    X_max = np.nanmax(X_train)
    X_min = np.nanmin(X_train)
    logX_scaled = (logX-X_min)/(X_max-X_min)-0.5
    return logX_scaled
```

```python
def log_min_max_scale_inverse(X,X_pred,select_idx):
    logX = np.log(X)
    X_train = logX[:,:,select_idx]
    X_max = np.nanmax(X_train)
    X_min = np.nanmin(X_train)
    X_pred_scale_inverse = (X_pred+0.5)*(X_max-X_min)+X_min
    return np.exp(X_pred_scale_inverse)
```

```python
T = scipy.io.loadmat('../../data/QoI3')['Tmax']
print("T shape:", T.shape)
```

```python
L = scipy.io.loadmat('../../data/QoI3')['L']
L = L[:,0]
print("L shape:", L.shape)
```

```python
select_idx = scipy.io.loadmat('../../data/training_idx')['idx_miniMax']# idx_random; idx_sparse; idx_Maxmin; idx_miniMax
select_idx = select_idx[:,0]
print("select_idx shape:", select_idx.shape)
```

```python
T_scaled = log_min_max_scale(X = T,select_idx = select_idx)
print("T_scaled shape:", T_scaled.shape)
```

```python
T_scaled_pred = np.copy(T_scaled)
for m in range(1000):
    display(m)
    Model_upperdir = 'tanh_logTmax' +'_{}mem'.format(20) +'_{}rec'.format(10)  +'_miniMax' + \
                        '_{}budget'.format(100)+'_{}'.format(1)
    T_scaled_pred[:,:,m] = scipy.io.loadmat(Model_upperdir+'/pred_{}.mat'.format(m))['T_pred']
```

```python
T_pred = log_min_max_scale_inverse(T,T_scaled_pred,select_idx)
```

```python
err = np.zeros((1000,))
for m in range(1000):
    err[m] = np.linalg.norm(T_pred[:L[m],0,m]-T[:L[m],0,m])
print(np.mean(err))
print(np.max(err))
```

```python
scipy.io.savemat(Model_upperdir+'/Tmax_pred.mat', {"T_pred": T_pred, "err": err})
```
